**1. MNE**

MNE (Magnetoencephalography and Electroencephalography) adalah library Python yang digunakan untuk mengolah dan menganalisis data sinyal otak, seperti EEG dan MEG.

Kegunaan:
- Membaca dan mengelola data EEG/MEG.
- Melakukan preprocessing data sinyal otak.
- Memvisualisasikan sinyal EEG.
- Membantu analisis data pada penelitian neuroscience dan Brain Computer Interface (BCI).

**2. PyPREP**

PyPREP adalah library Python yang digunakan untuk melakukan preprocessing atau pembersihan data EEG sebelum data dianalisis lebih lanjut.
Kegunaan:
- Menghilangkan noise pada sinyal EEG.
- Mendeteksi channel EEG yang bermasalah.
- Memperbaiki kualitas data EEG.
- Menyiapkan data agar siap digunakan untuk analisis atau machine learning

**3. SciPy**

SciPy adalah library Python yang digunakan untuk komputasi ilmiah dan pengolahan data numerik.

Kegunaan:
- Melakukan perhitungan matematika.
- Analisis statistik.
- Pengolahan sinyal digital.
- Membantu proses analisis data dan machine learning.

**4. Weights & Biases (wandb)**

wandb adalah library yang digunakan untuk mencatat dan memantau proses pelatihan model machine learning.

Kegunaan:
- Menyimpan hasil eksperimen.
- Memantau nilai akurasi dan loss.
- Membandingkan beberapa model yang diuji.
- Membuat visualisasi hasil pelatihan secara otomatis.

**5. pyECG**

pyECG adalah library yang digunakan untuk pengolahan dan analisis sinyal ECG (Electrocardiogram), yaitu sinyal aktivitas listrik jantung.

Kegunaan:
- Membaca data ECG.
- Mendeteksi denyut jantung.
- Menganalisis sinyal jantung.
- Membantu penelitian di bidang kesehatan dan biomedis.
