# **PRAKTIKUM 1**

### **Langkah 1 - Buat Dataset Sederhana (XOR)**

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
```

### **Langkah 2 - Inisialisasi Bobot dan Bias**

```python
# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

### **Langkah 3 - Proses Training (Forward Pass, Backpropagation, dan Update Bobot)**

```python
# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")
```

### **Langkah 4 - Menampilkan Hasil Prediksi Akhir**

```python
# Output akhir
print("Prediksi:")
print(a2)
```

Berdasarkan perbandingan di atas, penggunaan 3 neuron pada hidden layer (hidden_size = 3) memberikan performa yang lebih baik daripada konfigurasi awal (hidden_size = 2). Hal ini terlihat dari nilai loss akhir yang jauh lebih kecil (0.0027 berbanding 0.0081) serta hasil prediksi yang lebih akurat dan mendekati target biner XOR (nilai 0 semakin mendekati 0 dan nilai 1 semakin mendekati 1).

#### **Menambahkan Fungsi Aktivasi ReLU pada Hidden Layer**

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 3
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi ReLU untuk hidden layer dan Sigmoid untuk output layer
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = relu(z1)                 # Menggunakan ReLU pada hidden layer
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)              # Tetap menggunakan Sigmoid pada output layer

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * relu_derivative(z1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

Berdasarkan hasil uji coba, penggunaan fungsi aktivasi ReLU pada hidden layer memberikan performa lebih baik daripada Sigmoid. Hal ini terlihat dari nilai loss akhir yang turun drastis hingga mencapai angka sekitar 0,0003, jauh lebih kecil dibandingkan Sigmoid yang berada di kisaran 0,0027. Selain itu, hasil prediksi model juga menjadi jauh lebih akurat dalam mendekati target angka nol dan satu. Peningkatan performa ini terjadi karena ReLU mampu mengalirkan sinyal pelatihan dengan lebih lancar tanpa mengalami kendala pelemahan sinyal yang biasa terjadi pada Sigmoid, sehingga pembaruan bobot bisa berjalan jauh lebih optimal.

# **PRAKTIKUM 2**

### **Langkah 1 - Import Library**

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import tensorflow as tf
```

### **Langkah 2 - Load Dataset dan Preprocessing**

```python
# Load dataset
iris = load_iris()
X = iris.data
y = iris.target.reshape(-1, 1)

# One-hot encoding
encoder = OneHotEncoder(sparse_output=False)
y = encoder.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

### **Langkah 3 - Bangun Model**

```python
# Bangun model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

### **Langkah 4 - Kompilasi dan Latih Model**

```python
# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
model.fit(X_train, y_train, epochs=50, batch_size=8)
```

### **Langkah 5 - Evaluasi Hasil**

```python
# Evaluasi
loss, acc = model.evaluate(X_test, y_test)
print(f"Akurasi: {acc}")
```

### **Tugas 2 - Mengubah Jumlah Neuron Hidden Layer**

```python
# Bangun model
model_tugas2 = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# Kompilasi
model_tugas2.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
model_tugas2.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Evaluasi
loss_tugas2, acc_tugas2 = model_tugas2.evaluate(X_test, y_test)
print(f"Akurasi dengan konfigurasi baru: {acc_tugas2}")
```

Berdasarkan hasil uji coba, konfigurasi awal dengan jumlah neuron hidden layer 10 dan 8 menghasilkan akurasi yang sedikit lebih tinggi, yaitu sekitar 96,67% dibandingkan konfigurasi baru dengan 16 dan 16 neuron yang memperoleh akurasi sekitar 93,33%. Meskipun konfigurasi baru menghasilkan nilai loss yang lebih kecil pada data latih, akurasi pada data uji sedikit menurun. Hal ini menunjukkan bahwa penambahan jumlah neuron yang terlalu banyak tidak selalu meningkatkan performa secara otomatis pada dataset yang relatif kecil dan sederhana seperti Iris, bahkan berpotensi mengalami overfitting.

### **Tugas 3 - Membandingkan Sigmoid vs ReLU pada Dataset Iris**

```python
# 1. Model dengan Sigmoid pada Hidden Layer
print("--- Model dengan Sigmoid ---")

# Bangun model Sigmoid
model_sigmoid = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='sigmoid', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='sigmoid'),
    tf.keras.layers.Dense(3, activation='softmax')
])
# Kompilasi model Sigmoid
model_sigmoid.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model Sigmoid
model_sigmoid.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Evaluasi model Sigmoid
loss_sig, acc_sig = model_sigmoid.evaluate(X_test, y_test)
print(f"Sigmoid -> Loss: {loss_sig:.4f}, Akurasi: {acc_sig:.4f}\n")

# 2. Model dengan ReLU pada Hidden Layer
print("--- Model dengan ReLU ---")

# Bangun model ReLU
model_relu = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# Kompilasi model ReLU
model_relu.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model ReLU
model_relu.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Evaluasi model ReLU
loss_rel, acc_rel = model_relu.evaluate(X_test, y_test)
print(f"ReLU -> Loss: {loss_rel:.4f}, Akurasi: {acc_rel:.4f}")
```

Berdasarkan hasil uji coba, penggunaan fungsi aktivasi ReLU pada hidden layer memberikan kualitas pembelajaran yang lebih baik dibandingkan Sigmoid. Hal ini terlihat dari nilai loss akhir ReLU yang jauh lebih rendah, yakni sekitar 0,3415, sedangkan Sigmoid masih berada di angka 0,6144. Meskipun kedua model sama-sama menghasilkan tingkat akurasi yang identik pada data uji yaitu 96,67%, nilai error yang lebih kecil pada ReLU menunjukkan bahwa model mampu mempelajari pola data dengan lebih optimal.
