# **PRAKTIKUM 1**

### **Langkah 1 - Buat Dataset Sederhana (XOR)**

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
```

### **Langkah 2 - Inisialisasi Bobot dan Bias**

```python
# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

### **Langkah 3 - Proses Training (Forward Pass, Backpropagation, dan Update Bobot)**

```python
# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")
```

### **Langkah 4 - Menampilkan Hasil Prediksi Akhir**

```python
# Output akhir
print("Prediksi:")
print(a2)
```

Berdasarkan perbandingan di atas, penggunaan 3 neuron pada hidden layer (hidden_size = 3) memberikan performa yang lebih baik daripada konfigurasi awal (hidden_size = 2). Hal ini terlihat dari nilai loss akhir yang jauh lebih kecil (0.0027 berbanding 0.0081) serta hasil prediksi yang lebih akurat dan mendekati target biner XOR (nilai 0 semakin mendekati 0 dan nilai 1 semakin mendekati 1).

#### **Menambahkan Fungsi Aktivasi ReLU pada Hidden Layer**

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 3
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi ReLU untuk hidden layer dan Sigmoid untuk output layer
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = relu(z1)                 # Menggunakan ReLU pada hidden layer
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)              # Tetap menggunakan Sigmoid pada output layer

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * relu_derivative(z1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

Berdasarkan hasil uji coba, penggunaan fungsi aktivasi ReLU pada hidden layer memberikan performa lebih baik daripada Sigmoid. Hal ini terlihat dari nilai loss akhir yang turun drastis hingga mencapai angka sekitar 0,0003, jauh lebih kecil dibandingkan Sigmoid yang berada di kisaran 0,0027. Selain itu, hasil prediksi model juga menjadi jauh lebih akurat dalam mendekati target angka nol dan satu. Peningkatan performa ini terjadi karena ReLU mampu mengalirkan sinyal pelatihan dengan lebih lancar tanpa mengalami kendala pelemahan sinyal yang biasa terjadi pada Sigmoid, sehingga pembaruan bobot bisa berjalan jauh lebih optimal.
